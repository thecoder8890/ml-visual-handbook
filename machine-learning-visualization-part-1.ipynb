```python
# ============================================================================
# Cell 1: Importing Required Libraries and Setting Up Environment
# ============================================================================
# This cell imports all necessary libraries for data visualization
# - pandas/numpy: Data manipulation and numerical operations
# - seaborn/matplotlib: Statistical plotting and basic visualization
# - plotly: Interactive, web-based visualization
# - warnings filter: Suppresses non-critical warnings for cleaner output

import pandas as pd 
import numpy as np 
import seaborn as sns 
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt 
%matplotlib inline 

# Configure pandas to display numeric values with 2 decimal places
pd.set_option('display.precision', 2)

# Set Seaborn style with beige background and remove grid
sns.set(rc={"axes.facecolor":"Beige" , "axes.grid" : False})
#plt.style.use('fivethirtyeight')

# Suppress warnings to keep output clean
import warnings
warnings.filterwarnings("ignore")
```

## Section 1: Data Loading and Initial Exploration

This section covers loading the dataset, performing initial data exploration, and understanding the basic characteristics of the data.


```python
# Load the Heart Disease UCI dataset
# The dataset contains patient data with various medical attributes
df = pd.read_csv('/kaggle/input/heart-disease-data/heart_disease_uci.csv')

# Remove missing values to ensure clean data for visualization
df = df.dropna()

# Display first 2 rows to inspect data structure
df.head(2)
```

## Section 2: Basic Scatter Plots

Scatter plots are fundamental for visualizing relationships between two continuous variables. This section demonstrates various scatter plot techniques using Matplotlib and Plotly.


```python
# Create a scatter plot examining the relationship between age and cholesterol
# Grouped by the top 4 most common chest pain types

# Extract the top 4 chest pain types by frequency
top_leagues = df['cp'].value_counts().nlargest(4).index
display(top_leagues)

# Create scatter plot with:
# - x-axis: Age
# - y-axis: Cholesterol levels
# - Color: Different chest pain types
plt.figure(figsize=(15, 6))
sns.scatterplot(x='age', y='chol', data=df[df['cp'].isin(top_leagues)], hue='cp')
plt.title('Age vs. Cholesterol for Top 4 Chest Pain')
plt.xlabel('Age')
plt.ylabel('Cholesterol')
plt.legend(title='Chest Pain Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Interactive Scatter Plot using Plotly Express
# This creates an interactive plot where users can:
# - Hover over points to see data values
# - Zoom and pan
# - Toggle legend items on/off
# - Download the plot as PNG

fig = px.scatter(df, 
                 x='chol',      # x-axis: Cholesterol levels
                 y='age',       # y-axis: Patient age
                 color='sex')   # Color by gender

# Set custom dimensions for better visibility
fig.update_layout(width=1200, height=600)

# Add descriptive title
fig.update_layout(title_text='Scatter Plot of Cholesterol vs. Age (colored by Sex)')

# Display the interactive plot
fig.show()
```

```python
# Box Plot (5-Number Summary) using Plotly
# A box plot shows:
# - Minimum value
# - First quartile (Q1)
# - Median (Q2)
# - Third quartile (Q3)
# - Maximum value
# - Outliers (points beyond 1.5 * IQR)

from plotly.offline import iplot

fig = px.box(x = df["age"], 
            labels={"x":"Age"},
            title="5-Number-Summary(Box Plot) of Age")
iplot(fig)
```

```python
# Advanced Scatter Plot with Multiple Aesthetics
# This visualization encodes multiple variables:
# - Position (x, y): Cholesterol and Age
# - Color: Chest pain type
# - Size: Oldpeak values (stroke-related attribute)
# - Hover info: Exercise-induced angina

import plotly.express as px

fig = px.scatter(df, 
                 x='chol',           # Cholesterol on x-axis
                 y='age',            # Age on y-axis
                 color='cp',         # Color by chest pain type
                 size = 'oldpeak',   # Size of points represents oldpeak values
                 size_max = 30,      # Maximum bubble size
                 hover_name = 'exang' # Tooltip shows exercise-induced angina
                )  

fig.update_layout(width=1800, height=600)

fig.update_layout(title_text='Scatter Plot of Cholesterol vs. Age (colored by cp)')

fig.show()
```

```python
# Multi-variable Scatter Plot with Enhanced Styling
# This plot uses multiple visual cues to represent different attributes:
# - Bubble size: Number of major vessels colored
# - Color: Chest pain type
# - Hover data: Additional ST depression information

fig = px.scatter(data_frame = df, 
                 x="age",           # Age on x-axis
                 y="chol",          # Cholesterol on y-axis
                 color="cp",        # Color by chest pain type
                 size='ca',         # Bubble size represents major vessels
                 hover_data=['oldpeak']  # Additional hover information
                )

# Customize layout with enhanced typography
fig.update_layout(title_text="<b> Cholesterol Vs Age <b>", 
                  titlefont={'size': 24, 'family':'Serif'},
                  width=1500, 
                  height=750,
                 )

fig.show()
```

```python
# Faceted Scatter Plot with Logarithmic Scale
# This advanced visualization:
# - Creates separate subplots for each chest pain type
# - Uses logarithmic x-axis for better data distribution
# - Maintains multiple aesthetics (color, size, hover)

import plotly.express as px

fig = px.scatter(df, 
                 x='chol', 
                 y='age', 
                 color='cp', 
                 size = 'oldpeak', 
                 size_max = 30, 
                 hover_name = 'exang',
                 facet_col = 'cp',      # Create separate columns for each cp type
                 log_x = True            # Use logarithmic scale on x-axis
                )  

fig.update_layout(width=1600, height=600)

fig.update_layout(title_text='Scatter Plot of Cholesterol vs. Age (colored by cp)')

fig.show()
```

## Section 3: Bar Charts and Categorical Visualizations

Bar charts are essential for comparing categorical data. This section demonstrates various bar chart configurations including stacked, grouped, and percentage-based visualizations.


**Understanding the `hover_name` parameter:**
- `hover_name='exang'` means that the values in the 'exang' column will be shown as tooltips when you hover over the data points on the scatter plot
- This is useful for providing additional information about each data point without cluttering the plot with labels
- It helps maintain visual clarity while enabling detailed exploration through user interaction

```python
# Simple Bar Chart
# This creates a bar chart where:
# - Each bar represents a unique age value
# - Bar height represents cholesterol level
# - Color differentiates by gender
# - Hover shows additional oldpeak information

fig=px.bar(df,
           x='age',                    # x-axis: Age
           y='chol',                   # y-axis: Cholesterol levels
           hover_data=['oldpeak'],     # Additional hover information
           color='sex',                # Color by gender
           height=400)
fig.show()
```

```python
# Stacked Bar Chart
# This visualization shows counts of different slope values stacked within each chest pain type
# Function aggregates data by grouping on 'cp' (chest pain) and 'slope' (ST slope type)

def generate_rating_df(df):
    # Group data by chest pain type and slope, counting occurrences
    rating_df = df.groupby(['cp', 'slope']).agg({'id': 'count'}).reset_index()
    # Remove zero counts
    rating_df = rating_df[rating_df['id'] != 0]
    # Rename columns for clarity
    rating_df.columns = ['cp', 'slope', 'counts']
    # Sort by slope for consistent ordering
    rating_df = rating_df.sort_values('slope')
    return rating_df

rating_df = generate_rating_df(df)

# Create bar chart with stacked mode
fig = px.bar(rating_df, x='cp', y='counts', color='slope')

# Display text labels on bars
fig.update_traces(textposition='auto',  
                  textfont_size=20)  

# Stack bars instead of grouping them side by side
fig.update_layout(barmode='stack')

fig.show()
```

```python
# Grouped Bar Chart
# Similar to the previous chart but with 'group' mode instead of 'stack'
# This allows direct comparison of slope values across chest pain types

def generate_rating_df(df):
    rating_df = df.groupby(['cp', 'slope']).agg({'id': 'count'}).reset_index()
    rating_df = rating_df[rating_df['id'] != 0]
    rating_df.columns = ['cp', 'slope', 'counts']
    rating_df = rating_df.sort_values('slope')
    return rating_df

rating_df = generate_rating_df(df)
fig = px.bar(rating_df, x='cp', y='counts', color='slope')

fig.update_traces(textposition='auto',  
                  textfont_size=20)  

# Group mode places bars side by side for easier comparison
fig.update_layout(barmode='group')

fig.show()
```

```python
# Grouped Bar Chart with Explicit Text Display
# This version explicitly specifies the 'text' parameter to show count values on bars

import plotly.express as px

def generate_rating_df(df):
    rating_df = df.groupby(['cp', 'slope']).agg({'id': 'count'}).reset_index()
    rating_df = rating_df[rating_df['id'] != 0]
    rating_df.columns = ['cp', 'slope', 'counts']
    rating_df = rating_df.sort_values('slope')
    return rating_df

rating_df = generate_rating_df(df)

# Create bar chart with grouped mode and count labels
fig = px.bar(rating_df, 
             x='cp', 
             y='counts', 
             color='slope', 
             barmode='group', 
             text='counts'  # Explicitly display count values
             )  

fig.update_traces(textposition='auto', 
                  textfont_size=20)  

fig.show()
```

```python
# Grouped Bar Chart with Percentage Labels
# This advanced bar chart shows both count values and their percentages
# Useful for understanding relative proportions within each category

def generate_rating_df(df):
    rating_df = df.groupby(['cp', 'slope']).agg({'id': 'count'}).reset_index()
    rating_df = rating_df[rating_df['id'] != 0]
    rating_df.columns = ['cp', 'slope', 'counts']
    rating_df = rating_df.sort_values('slope')
    
    # Calculate percentage of each group relative to total
    total_counts = rating_df['counts'].sum()
    rating_df['percentage'] = rating_df['counts'] / total_counts * 100
    
    return rating_df

rating_df = generate_rating_df(df)

# Create bar chart showing counts but display percentage labels
fig = px.bar(rating_df, x='cp', y='counts', color='slope', text='percentage')

# Format percentage text with 1 decimal place
fig.update_traces(
    texttemplate='%{text:.1f}%',  # Format: XX.X%
    textposition='outside',       # Place labels outside bars
    textfont_size=16              
)

# Update axes labels for clarity
fig.update_layout(
    barmode='group',
    yaxis_title='Count',
    xaxis_title='CP',
    legend_title='Slope'
)

# Customize dimensions and title
fig.update_layout(
    height=550,  
    width=1500,   
    title_text="Distribution of Chest Pain Type by Percentage",
    title_font_size=24
)

fig.show()
```

## Section 4: Marginal Distributions and Advanced Plots

This section explores marginal distribution plots (combining scatter plots with histograms/box plots) and advanced visualization techniques.


```python
# Marginal Distribution Plot
# Combines scatter plot with marginal distributions to show:
# - Main plot: Scatter of age vs cholesterol
# - Top margin: Histogram of age distribution
# - Right margin: Box plot of cholesterol distribution

fig = px.scatter(data_frame = df, 
                 x="age",                      # x-axis: Age
                 y="chol",                     # y-axis: Cholesterol
                 color="cp",                   # Color by chest pain type
                 size='ca',                    # Bubble size by major vessels
                 hover_data=['oldpeak'],       # Additional hover info
                 marginal_x="histogram",       # Top marginal: Age histogram
                 marginal_y="box"              # Right marginal: Cholesterol box plot
                )

fig.update_layout(title_text="<b> Age vs Cholesterol <b>", 
                  titlefont={'size': 24, 'family':'Serif'},
                  width=1500, 
                  height=750,
                 )

fig.show()
```

```python
# Scatter Plot with Ordinary Least Squares (OLS) Trendline
# Adds a regression line to show the relationship trend between age and cholesterol
# The trendline helps identify overall patterns and correlations

fig = px.scatter(data_frame = df, 
                 x="age", 
                 y="chol", 
                 size ="ca",
                 size_max=30,
                 color= "sex",
                 trendline="ols")    # Add OLS regression line to each sex group

fig.update_layout(title_text="<b> Age vs Cholesterol <b>", 
                  titlefont={'size': 24, 'family':'Serif'},
                  width=1500, 
                  height=750,
                 )

fig.show()
```

```python
# Scatter Plot with Overall OLS Trendline
# Similar to previous but draws a single trendline across all data points (not per group)
# Useful for analyzing the overall relationship regardless of category

fig = px.scatter(data_frame = df, 
                 x="age", 
                 y="chol", 
                 size ="ca",
                 size_max=30,
                 color= "sex",
                 trendline="ols",
                 trendline_scope="overall",        # Single trendline for all data
                 trendline_color_override="black"  # Override default color to black
                )

fig.update_layout(title_text="<b>Chest Pain vs Gender<b>", 
                  titlefont={'size': 24, 'family':'Serif'},
                  width=1500, 
                  height=750,
                 )
fig.show()
```

```python
# Outlier Detection and Grouped Bar Chart
# This cell demonstrates:
# 1. Calculating IQR (Interquartile Range) for outlier detection
# 2. Counting outliers in cholesterol levels
# 3. Creating grouped bar charts comparing age and cholesterol by chest pain type

import numpy as np
import plotly.graph_objs as go
from plotly.offline import iplot

# Calculate quartiles and IQR for cholesterol
Q25 = np.quantile(df['chol'], q=0.25)  # First quartile (25th percentile)
Q75 = np.quantile(df['chol'], q=0.75)  # Third quartile (75th percentile)
IQR = Q75 - Q25
cut_off = IQR * 1.5                    # Standard outlier threshold

# Print number of outliers in cholesterol data
print('Number of Cholesterol Lower Outliers:', df[df['chol'] <= (Q25 - cut_off)]['chol'].count())
print('Number of Cholesterol Upper Outliers:', df[df['chol'] >= (Q75 + cut_off)]['chol'].count())

# Group by chest pain type and aggregate by summing
temp = df.groupby('cp').sum().sort_values('age', ascending=False)

# Create grouped bar chart showing age and cholesterol totals
data = [
    go.Bar(x=temp.index, y=temp['age'], name='Age', text=temp['age'], textposition='auto'),
    go.Bar(x=temp.index, y=temp['chol'], name='Cholesterol', text=temp['chol'], textposition='auto')
]

# Define layout with large, readable fonts
layout = go.Layout(
    xaxis=dict(title='Chest Pain', titlefont=dict(size=25)),
    yaxis=dict(title='Values', titlefont=dict(size=25)),
    showlegend=True,
    width=1300,
    height=600
)

# Create and display figure
fig = go.Figure(data=data, layout=layout)
iplot(fig)
```

```python
import pandas as pd
import plotly.express as px

# Assuming you have the gap_df DataFrame loaded

# Calculate the IQR for both 'gdp_cap' and 'life_exp'
Q1_gdp_cap = gap_df['gdp_cap'].quantile(0.25)
Q3_gdp_cap = gap_df['gdp_cap'].quantile(0.75)
IQR_gdp_cap = Q3_gdp_cap - Q1_gdp_cap

Q1_life_exp = gap_df['life_exp'].quantile(0.25)
Q3_life_exp = gap_df['life_exp'].quantile(0.75)
IQR_life_exp = Q3_life_exp - Q1_life_exp

# Filter out outliers based on IQR
gap_df_no_outliers = gap_df[(gap_df['gdp_cap'] >= Q1_gdp_cap - 1.5 * IQR_gdp_cap)
                            & (gap_df['gdp_cap'] <= Q3_gdp_cap + 1.5 * IQR_gdp_cap)
                            & (gap_df['life_exp'] >= Q1_life_exp - 1.5 * IQR_life_exp)
                            & (gap_df['life_exp'] <= Q3_life_exp + 1.5 * IQR_life_exp)]

# Create the scatter plot with the outlier-filtered data
fig = px.scatter(gap_df_no_outliers, x='gdp_cap', y='life_exp', color='continent',
                 size='population', size_max=60, hover_name="country")

fig.update_layout(
    height=550,  
    width=1500,   
    title_text="Distribution of GDP Cap Vs Life Exp without Outliers",
    title_font_size=24
)

fig.show()
```

```python
# Salary Data Cleaning Function
# Converts salary strings with ranges to numeric mean values
# Handles various formats: currency symbols, ranges, abbreviations
# Returns mean of salary range or None if invalid

import re

def clean_and_calculate_mean(salary):
    try:
        # Remove currency symbols, words, and extra characters
        salary = salary.replace('RM', '').replace('MYR', '').replace('$', '')
        salary = salary.replace('per month', '').replace('p.m.', '').replace('per year', '').replace('p.a.', '').strip()

        # Handle ranges with different separators (various dash types)
        if '–' in salary:
            salary_range = salary.split('–')
        elif '-' in salary:
            salary_range = salary.split('-')
        elif '—' in salary:
            salary_range = salary.split('—')
        else:
            salary_range = [salary]

        # Convert values to integers, handling potential errors
        salary_values = []
        for value in salary_range:
            try:
                value = int(float(value.replace(',', '').strip()))
                salary_values.append(value)
            except ValueError:
                pass  # Ignore non-numeric values

        # Calculate mean if at least two valid values are found
        if len(salary_values) >= 2:
            salary_mean = sum(salary_values) / len(salary_values)
            return salary_mean
        else:
            return None

    except Exception as e:
        print(f"Error processing salary '{salary}': {e}")
        return None

# Apply the function to clean salary data
job['Salary'] = job['salary'].apply(clean_and_calculate_mean)
job = job.drop('salary', axis=1)  # Remove original salary column
job.head(2)
```

```python
# Outlier Detection and Removal for Salary Data
# Compares salary distributions before and after outlier removal
# Helps understand the impact of outliers on data analysis
# Uses IQR (Interquartile Range) method for outlier detection

def remove_outliers_iqr(df, column):
    # Calculate quartiles (using 10th and 85th percentiles for stricter filtering)
    Q1 = df[column].quantile(0.10)
    Q3 = df[column].quantile(0.85)
    
    # Calculate IQR
    IQR = Q3 - Q1
    
    # Define outlier boundaries
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Filter out outliers
    df_outlier_free = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    
    return df_outlier_free

# Remove outliers from salary data
jobs = remove_outliers_iqr(job, 'Salary')

# Get top 20 job titles from both datasets (original and outlier-removed)
top_leagues_job = job['job_title'].value_counts().nlargest(20).index
top_leagues_jobs = jobs['job_title'].value_counts().nlargest(20).index

# Combine the top job titles
top_leagues = top_leagues_job.union(top_leagues_jobs)

# Left plot: Original data with outliers
plt.figure(figsize=(16, 6))
sns.boxplot(x='Salary', y='job_title', data=job[job['job_title'].isin(top_leagues)])
plt.title('Distribution of Salary by Top 20 job_title from job')
plt.xlabel('Salary')
plt.ylabel('job_title')
plt.show()

# Right plot: Data after removing outliers
plt.figure(figsize=(16, 6))
sns.boxplot(x='Salary', y='job_title', data=jobs[jobs['job_title'].isin(top_leagues)])
plt.title('Distribution of Salary by Top 20 job_title from After Removing Outliers')
plt.xlabel('Salary')
plt.ylabel('job_title')
plt.show()
```
