## 3️⃣ Your First 3D Scatter Plot

### What is a 3D Scatter Plot?

A 3D scatter plot extends the traditional 2D scatter plot by adding a third axis (Z). This allows us to visualize three variables simultaneously:
- **X-axis**: First variable
- **Y-axis**: Second variable
- **Z-axis**: Third variable

### When to Use 3D Plots:

✅ **Good Use Cases:**
- Exploring relationships between exactly 3 variables
- Interactive visualizations where users can rotate the view
- Presenting to audiences who can interact with the plot
- When 3D structure genuinely exists in the data

❌ **Avoid 3D When:**
- Printing to static documents (use multiple 2D plots instead)
- The third dimension doesn't add meaningful information
- Audience can't interact with the plot
- 2D alternatives would be clearer

### Reading 3D Plots:
- **Rotate**: Click and drag to change viewing angle
- **Zoom**: Scroll to zoom in/out
- **Pan**: Right-click and drag to move the view

```python
# Create a 3D scatter plot with Wine Quality data
if wine is not None and len(wine) > 0:
    print("🎨 Creating 3D scatter plot: Alcohol vs pH vs Quality\n")
    
    fig = plt.figure(figsize=(12, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    # Create scatter plot
    # Color by quality for better visualization
    scatter = ax.scatter(
        wine['alcohol'] if 'alcohol' in wine.columns else wine.iloc[:, 0],
        wine['pH'] if 'pH' in wine.columns else wine.iloc[:, 1],
        wine['quality'] if 'quality' in wine.columns else wine.iloc[:, -1],
        c=wine['quality'] if 'quality' in wine.columns else wine.iloc[:, -1],
        cmap='viridis',
        s=50,
        alpha=0.6,
        edgecolors='black',
        linewidth=0.5
    )
    
    # Labels and title
    ax.set_xlabel('Alcohol Content (%)', fontsize=12, labelpad=10)
    ax.set_ylabel('pH Level', fontsize=12, labelpad=10)
    ax.set_zlabel('Quality Score', fontsize=12, labelpad=10)
    ax.set_title('3D Scatter: Wine Characteristics', fontsize=14, fontweight='bold', pad=20)
    
    # Add color bar
    cbar = plt.colorbar(scatter, ax=ax, pad=0.1, shrink=0.8)
    cbar.set_label('Quality Score', rotation=270, labelpad=20)
    
    # Improve viewing angle
    ax.view_init(elev=20, azim=45)
    
    plt.tight_layout()
    plt.show()
    
    print("\n📊 Interpretation Guide:")
    print("- Each point represents one wine sample")
    print("- Color intensity = Quality score (darker = better)")
    print("- Look for clusters or patterns in 3D space")
    print("- Higher quality wines may cluster in specific regions")
else:
    print("⚠️ Wine dataset not available for 3D plot")
```

## 4️⃣ Interactive 3D Plots with Plotly

### Why Plotly for 3D?

Plotly creates **interactive 3D visualizations** that users can:
- Rotate freely with mouse
- Zoom in/out smoothly
- Hover to see exact values
- Toggle data series on/off
- Export as static images

### Interactive Features:
- **Hover**: See data point details
- **Click & Drag**: Rotate the 3D view
- **Scroll**: Zoom in and out
- **Camera Controls**: Reset view, toggle projection

Let's create an interactive 3D scatter plot showing the relationship between GDP, life expectancy, and population.

```python
# Filter for a specific year to reduce clutter
df_2007 = df_gapminder[df_gapminder['year'] == 2007]

print(f"📊 Creating interactive 3D scatter plot ({len(df_2007)} countries)\n")

# Create interactive 3D scatter plot
fig = px.scatter_3d(
    df_2007,
    x='gdpPercap',  # X-axis: GDP per capita
    y='lifeExp',     # Y-axis: Life expectancy
    z='pop',         # Z-axis: Population
    color='continent',  # Color by continent
    size='pop',      # Bubble size by population
    hover_name='country',  # Show country on hover
    hover_data={
        'gdpPercap': ':$,.0f',
        'lifeExp': ':.1f',
        'pop': ':,'
    },
    title='Global Development Indicators in 3D (2007)',
    labels={
        'gdpPercap': 'GDP per Capita (USD)',
        'lifeExp': 'Life Expectancy (years)',
        'pop': 'Population',
        'continent': 'Continent'
    },
    height=700,
    size_max=30,
    opacity=0.7
)

# Customize layout
fig.update_layout(
    scene=dict(
        xaxis=dict(type='log', title='GDP per Capita (log scale)'),
        yaxis=dict(title='Life Expectancy'),
        zaxis=dict(type='log', title='Population (log scale)'),
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=1.3)  # Initial viewing angle
        )
    ),
    font=dict(size=12)
)

fig.show()

print("\n🎯 What to Explore:")
print("1. Rotate the plot to see different angles")
print("2. Hover over points to identify countries")
print("3. Notice how continents cluster in different regions")
print("4. Observe the relationship: higher GDP ↔ higher life expectancy")
print("5. See how population (bubble size) varies")
print("\n💡 Tip: Use the camera controls in the top right!")
```

### Understanding Log Scales in 3D

In the plot above, we used **logarithmic scales** for GDP and Population. Here's why:

**Why Use Log Scales?**
- GDP ranges from $400 to $100,000+ (250x difference)
- Population ranges from 10,000 to 1,000,000,000+ (100,000x difference)
- Linear scales would squash most data into a corner
- Log scales spread data evenly across the range

**Reading Log Scales:**
- Each major tick is 10x the previous (100, 1000, 10000...)
- Equal distances = equal proportional changes
- Example: 100→1000 is same distance as 1000→10000

## 5️⃣ Animated 3D Visualizations

### Time-Series in 3D

We can add a **fourth dimension** (time) to 3D plots through animation. This shows how 3D relationships evolve over time.

**Benefits:**
- See trends and changes
- Identify which countries develop faster
- Track movements in 3D space
- Tell stories with data

**How to Use:**
- Click Play to start animation
- Use slider to jump to specific years
- Pause to examine interesting moments
- Rotate while animated for different perspectives

```python
print("🎬 Creating animated 3D visualization...\n")

# Create animated 3D scatter
fig_animated = px.scatter_3d(
    df_gapminder,
    x='gdpPercap',
    y='lifeExp',
    z='pop',
    color='continent',
    size='pop',
    hover_name='country',
    animation_frame='year',  # THIS CREATES THE ANIMATION
    animation_group='country',  # Group by country for smooth transitions
    hover_data={
        'gdpPercap': ':$,.0f',
        'lifeExp': ':.1f',
        'pop': ':,'
    },
    title='Global Development Over Time (1952-2007)',
    labels={
        'gdpPercap': 'GDP per Capita',
        'lifeExp': 'Life Expectancy',
        'pop': 'Population'
    },
    range_x=[100, 100000],  # Fixed ranges for consistent animation
    range_y=[20, 90],
    range_z=[0, 1.5e9],
    height=700,
    size_max=30,
    opacity=0.7
)

# Set log scales and camera angle
fig_animated.update_layout(
    scene=dict(
        xaxis=dict(type='log'),
        yaxis=dict(title='Life Expectancy'),
        zaxis=dict(type='log'),
        camera=dict(eye=dict(x=1.5, y=1.5, z=1.3))
    )
)

# Adjust animation speed
fig_animated.layout.updatemenus[0].buttons[0].args[1]['frame']['duration'] = 1000
fig_animated.layout.updatemenus[0].buttons[0].args[1]['transition']['duration'] = 500

fig_animated.show()

print("\n📈 Historical Trends to Watch:")
print("1. Overall upward movement (increasing life expectancy)")
print("2. Rightward movement (increasing GDP)")
print("3. Some countries move faster than others")
print("4. Asia's rapid development in recent decades")
print("5. Population growth (increasing bubble sizes)")
```

## 6️⃣ 3D Surface Plots

### What Are Surface Plots?

Surface plots show a **continuous surface** in 3D space, representing how Z changes as a function of X and Y.

**Use Cases:**
- Mathematical functions: z = f(x, y)
- Terrain elevation maps
- Temperature/pressure fields
- Probability density surfaces
- Response surfaces in experiments

**Reading Surface Plots:**
- **Height**: Z-value (the function output)
- **Color**: Often represents Z-value or gradient
- **Peaks**: Maximum values
- **Valleys**: Minimum values
- **Ridges**: Lines of high values

Let's create a mathematical surface to demonstrate:

```python
print("🏔️ Creating 3D surface plot\n")

# Create a mathematical surface: z = sin(√(x² + y²))
# This creates a "ripple" effect radiating from the origin

# Create grid
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Calculate Z values
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R) / (R + 0.1)  # Add small constant to avoid division by zero

# Create surface plot with Plotly for interactivity
fig = go.Figure(data=[go.Surface(
    x=X,
    y=Y,
    z=Z,
    colorscale='Viridis',
    contours={
        "z": {"show": True, "usecolormap": True, "highlightcolor": "limegreen", "project": {"z": True}}
    }
)])

fig.update_layout(
    title='3D Surface Plot: Ripple Function',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z = sin(√(x²+y²)) / (√(x²+y²) + 0.1)',
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=1.3)
        )
    ),
    height=700
)

fig.show()

print("\n🔍 What You're Seeing:")
print("- Central peak where x=0, y=0")
print("- Ripples radiating outward (like water waves)")
print("- Contour lines projected on bottom (like topographic map)")
print("- Color gradient shows height (purple=low, yellow=high)")
print("\n💡 Rotate the plot to see the structure from different angles!")
```
